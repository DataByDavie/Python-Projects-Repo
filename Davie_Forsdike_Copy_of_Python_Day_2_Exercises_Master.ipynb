### 1 ) 🔐 Scenario-Based Exercise: Security Door Access
Scenario:
You are a security engineer at a high-tech facility. To enter the building, employees must type in a secret code on a keypad. The code for today is "123". You need to write a program that checks if the person has entered the correct code.

🧠 Mission Brief:
Your task is to write a program that:

Asks the person to enter the access code.

If the code is "123", the system should print:
"🔓 Access granted. Welcome inside!"

If the code is incorrect, the system should print:
"⛔ Access denied. Incorrect code."



```python
code = 123                                                #This sets the correct code to 123
user_code = int(input("Please enter today's code:"))      #And prompts the user to input today's code

while user_code != code:                                  #Then it uses a While loop to checkk if the user's code does not match the
  print("⛔ Access denied. Incorrect code.")              #correct code. If it doesn't match, then it displays the error code and prompts
  user_code = int(input("Please enter today's code:"))    #the user to try again

  if user_code == code:                                   #Then it uses an IF statement to check if the newly entered code matches.
    print("🔓 Access granted. Welcome inside!")           #If it does, then the success message is sent and the loop is broken.
    break
```

### 3) You're developing a simple calculator tool for a local community learning website. Users should be able to choose what operation they want to perform and continue using the calculator until they decide to exit.

🧪 Task:
Write a Python program that:

Displays a menu with the following options:

* 1- Addition

* 2- Subtraction

* 3- Multiplication

* 4-Division

* 5- Exit

Asks the user to choose an operation

Asks for two numbers, then performs the chosen operation

Repeats the menu until the user selects Exit

```python
while True:

  print("Choose your operation below: \n1. Addition \n2. Subtraction \n3. Multiplication \n4. Division \n5. Exit")

  choice = input("Please choose an operation number (1-5): ")         #While true keeps the calculator running until option 5 is chosen
  if choice == '5':                                                   #It checks if option 5 is chosen before going into the operations for
    print(f"Exiting Calculator")                                      #each loop.
    break

  if choice in ('1', '2', '3', '4'):                                  #Then it checks which option is chosen
    num1 = int(input("Please enter your 1st number: "))               #And the user inputs the two numbers they want for that operation
    num2 = int(input("Please enter your 2nd number: "))

    if choice == '1':                                                 #Depending on which operation the user selects, the program flows
      answer = num1 + num2                                            #to the correct if statement and dispalys the results
      print(f"Answer: {num1} + {num2} = {answer}")

    elif choice == '2':
      answer = num1 - num2
      print(f"Answer: {num1} - {num2} = {answer}")

    elif choice == '3':
      answer = num1 * num2
      print(f"Answer: {num1} * {num2} = {answer}")

    elif choice == '4':                                               #For division, it has a logic check to prevent users dividing by 0
      if num2 == 0:
        print("You can't divide by zero")
      else:
        answer = num1 / num2
        print(f"Answer: {num1} / {num2} = {answer}")
else:
    print("Please choose an operation between 1 and 5.")             #If an option other than 1-5 is chosen then it prompts the message
```

### 4) You’re building a fitness tracker app. The app should calculate the total number of steps someone would take if they increase their steps by 1 each day — starting from 1 up to a number they choose.

🧪 Task:
Write a Python program that:

Asks the user to enter a positive number

Calculates the sum of all numbers from 1 to that number

Displays the total
<p> If the user enters 10, the output should be:

"The sum of numbers from 1 to 10 is 55"

```python
number = int(input("Please enter a positive number: "))     #Takes an input from the user

while number <= 0:                                          #While loop repeats the prompt until a number above 0 is entered
  print("Please enter a value above 0")
  number = int(input("Please enter a positive number: "))

steps = 0                                                   #This sets the step counter to 0 from the start
for i in range(1, number + 1):                              #This For loop iterates through every number from 1 up to the inputted number
    steps += i                                              #On each iteration, the current number (i) is added to the step count variable
                                                            #in the 'steps =+ i' shortcut expression to say steps = steps + i.
print(f"The sum of numbers from 1 to {number} is: {steps}")

```

### 5) You’ve been asked to help design a feature for a Math Quiz App. One challenge asks the user to enter a number — and your program will check whether it’s a prime number (a number that has only two factors: 1 and itself).

🧪 Task:
Write a Python program that:

Asks the user to enter a positive integer

Checks whether the number is a prime number

Displays a message telling the user whether the number is prime or not



```python
number = int(input("Please enter a positive number: "))     #Takes an input from the user

while number <= 0:                                          #While loop repeats the prompt until a number above 0 is entered
    print("Please enter a value above 0")
    number = int(input("Please enter a positive number: "))

if number < 2:                                             #1 is not a prime number but the code would've returned it as one
    print(f"{number} is NOT a prime number")
else:
    prime = True                                            #This set's the boolean of the prime variable as true from the start
    for i in range(2, int(number**0.5) + 1):                #Then it uses a For loop to test every number between 2 and the square root
        if number % i == 0:                                 #of the inputted number. Then it modulus divides all the numbers in the range
            prime = False                                   #and if any leaves a remainder of zero, then the prime status is set to false.
            break

    if prime:
        print(f"{number} IS a prime number")
    else:
        print(f"{number} is NOT a prime number")
```

### 6) You’re building a tool for a puzzle game where certain levels require calculating the factorial of a number to unlock a clue. A factorial of a number is the product of all positive integers from 1 up to that number.

For example:
5! = 5 × 4 × 3 × 2 × 1 = 120

🧪 Task:
Write a Python program that:

Asks the user to enter a non-negative integer

Calculates the factorial of that number

Displays the result



```python
number = int(input("Please enter a positive integer: ")) #Takes an input from the user

while number < 0:                                                    #Makes sure the number is above 0 so the factorial can start properly
    print("Please enter a value of 0 or higher.")
    number = int(input("Please enter a positive integer: ")) #Repeats until a number abover 0 is entered

factorial = 1                                                        #Sets the factorial to 1 so that the multiplication chain can start

for i in range(1, number + 1):           #Uses a For loop to iterate through every number from 1 to the inputted number
    factorial *= i                       #Each loop the next number in the sequence is multiplied into the current number of the factorial

print(f"The factorial of {number} is: {factorial}")
```
